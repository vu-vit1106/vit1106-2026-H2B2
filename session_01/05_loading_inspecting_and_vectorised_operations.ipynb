## What is a CSV file?

CSV means comma-separated values.

A CSV file is a plain text file that stores table data. Each line is one row. Commas separate the values.

Pandas can read a CSV file and turn it into a `DataFrame`.


## Import pandas

```python
import pandas as pd

print("pandas is ready")
```

## Load the dataset

The dataset for this activity is stored in the `data` folder for this session.

The file path is `data/student_results.csv`.

```python
from pathlib import Path
DATA_FOLDER = Path("data")
STUDENT_RESULTS_FILE = DATA_FOLDER / "student_results.csv"

results = pd.read_csv(STUDENT_RESULTS_FILE)

results
```

## Vectorised operations

A vectorised operation works on a whole column at once.

This is different from writing a loop for each row. It is also different from typing a spreadsheet formula such as `=A2+5` into one cell and copying it down.

The next cell uses one pandas expression, `results["score"] + 5`, to add 5 bonus marks to every score.


```python
results["score"] + 5
```

## Save the result in a new column

A calculation is only temporary unless we store it.

The next cell stores the result of `results["score"] + 5` in a new column.


```python
results["score_with_bonus"] = results["score"] + 5

results.head()
```

## Vectorised comparisons

A comparison can also work across a whole column.

The next cell checks whether each student passed by using `results["score"] >= 50`.


```python
results["score"] >= 50
```

## Store a comparison in a new column

The next cell stores the `True` or `False` result in a new column named `passed`.


```python
results["passed"] = results["score"] >= 50

results.head()
```

## Filtering with a vectorised comparison

We can use a `True` or `False` column, or a direct condition, to filter rows.


```python
results[results["passed"] == True]
```

Grouping lets us compare categories.

The next cell calculates the average `score` for each `campus` using `.groupby()` and `.mean()`.

## Sorting a dataset

Sorting helps us find high or low values quickly.

```python
results.sort_values("score", ascending=False)
```
