## From dictionaries to tables

In the previous notebook, you used a dictionary of lists.

That structure already looks like a table:

- each key is like a column name
- each list is like a column of values
- each position in the lists is like one row

Pandas can turn that structure into a DataFrame.

````python
student_table = {
    "student": ["Taylor Nguyen", "Mia Chen", "Jordan Lee"],
    "campus": ["Footscray", "City", "Footscray"],
    "score": [82, 74, 46]
}

print(student_table)
````

## Import pandas and NumPy

Pandas is the Python library we use for `DataFrame` and `Series` objects.

NumPy is included here so we can compare a Python list, a NumPy array, and a pandas `Series` side by side.

The common short names are `pd` for pandas and `np` for NumPy. Many examples online use these same short names.


````python
import pandas as pd
import numpy as np

print("pandas is ready")
print(pd.__version__)
````

## Lists, arrays, and Series

Before we create a full `DataFrame`, it helps to compare three one-dimensional structures.

| Structure | Main idea | Typical use |
|---|---|---|
| Python list | A built-in sequence of values | Store general values in order |
| NumPy array | A fast numeric sequence | Do calculations on many numbers at once |
| pandas `Series` | One labelled column of data | Work with one column before it becomes part of a `DataFrame` |

A `Series` is like one column from a table. It has values, and it can also have a label called a name.


````python
scores_list = [78, 45, 92]
scores_array = np.array(scores_list)
scores_series = pd.Series(scores_list, name="score")

print("Python list:")
print(scores_list)
print(type(scores_list))

print("\nNumPy array:")
print(scores_array)
print(type(scores_array))

print("\npandas Series:")
print(scores_series)
print(type(scores_series))
````

## Series labels

A `Series` can also use an index. The index gives each value a label.

This is useful in data analysis because table values usually mean something. A score is more useful when we know which student it belongs to.


````python
student_scores = pd.Series(
    [78, 45, 92],
    index=["Taylor Nguyen", "Mia Chen", "Jordan Lee"],
    name="score"
)

student_scores
````

## From Series to DataFrames

A `DataFrame` is a table. One simple way to think about it is:

- a `Series` is one labelled column
- a `DataFrame` is several labelled columns placed side by side

The dictionary of lists below can be turned into a `DataFrame` because each list becomes one column.


## Create a DataFrame

The next cell turns `student_table` into a DataFrame.

In Jupyter, writing the DataFrame name on the last line displays it as a table.

````python
df_students = pd.DataFrame(student_table)

df_students
````

## Reading the table shape

The shape tells us how many rows and columns the DataFrame has.

The first number is rows.

The second number is columns.

````python
print(df_students.shape)
print("Rows:", df_students.shape[0])
print("Columns:", df_students.shape[1])
````

## Column names

A DataFrame has column labels.

These labels are important because we use them to select, calculate, filter, and organise data.

````python
print(df_students.columns)
````

## Selecting one column

Use square brackets with the column name to select one column.

One column in pandas is a `Series`. It is the same idea you saw earlier: one labelled column of values.

When the column comes from a `DataFrame`, pandas keeps the row labels with it.


````python
df_students["score"]
````

## Selecting several columns

Use a list of column names to select more than one column.

Notice the double square brackets: one pair for selecting, and one pair for the list of column names.

````python
df_students[["student", "score"]]
````
