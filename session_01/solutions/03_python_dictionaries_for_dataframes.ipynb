```python
def clean_text(text):
    return text.strip().title()

def student_summary(record):
    name = record["name"]
    score = record["score"]
    return f"{name} has a score of {score}."

def pass_message(record):
    if record["score"] >= 50:
        return record["name"] + " has passed."
    return record["name"] + " has not passed yet."

print(student_summary(student))
print(pass_message(student))
```

```python
students = [
    {"name": "Taylor Nguyen", "campus": "Footscray", "score": 82},
    {"name": "Mia Chen", "campus": "City", "score": 74},
    {"name": "Jordan Lee", "campus": "Footscray", "score": 46}
]

for record in students:
    print(record["name"], record["campus"])

total = 0
for record in students:
    total = total + record["score"]
average = total / len(students)
print(average)
```

```python
student_table = {
    "name": ["Taylor Nguyen", "Mia Chen", "Jordan Lee"],
    "campus": ["Footscray", "City", "Footscray"],
    "score": [82, 74, 46]
}

student_position = 2
print(student_table["name"][student_position])
print(student_table["campus"][student_position])
print(student_table["score"][student_position])
```

```python
def get_student(n):
    return {
        "name": student_table["name"][n],
        "campus": student_table["campus"][n],
        "score": student_table["score"][n]
    }

print(get_student(2))
```

```python
practice_students = [
    {"name": "  priya shah", "score": 69},
    {"name": "  noah williams", "score": 51},
    {"name": "  liam brown", "score": 43}
]

for record in practice_students:
    record["name"] = clean_text(record["name"])
    print(record["name"])
    print(record["score"])
    print(pass_message(record))
```
