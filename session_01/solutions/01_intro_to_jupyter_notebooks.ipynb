## Why lists can become awkward

Lists work well for small groups of values.

But if we need to do the same calculation to every value, the code can become repetitive.

For example, the code below adds 5 bonus marks to each mark.

```python
bonus_marks = []

for mark in marks:
    bonus_marks.append(mark + 5)

print(bonus_marks)
```

This works, but it takes several lines. When we work with lots of numbers, we often want a shorter and clearer way.

That is one reason we will later use NumPy arrays.

## List comprehensions

A list comprehension creates a new list from an existing list.

It is useful when you want to do the same simple action to each value.

The example below does the same work as the previous loop.

```python
bonus_marks = [mark + 5 for mark in marks]

print(bonus_marks)
```

Read it from left to right:

- `mark + 5` is the new value we want
- `for mark in marks` means do it for each value in the list

The original list is not changed. A new list is created.

## Task 11: Add to each value

Use a list comprehension to create a new list called `extra_hours`.

It should add 1 to each value in your `weekly_hours` list.

Print the new list.

```python
extra_hours = [hours + 1 for hours in weekly_hours]

print(extra_hours)
```

## Filtering with a list comprehension

A list comprehension can also keep only some values.

The example below keeps marks that are 70 or higher.

```python
high_marks = [mark for mark in marks if mark >= 70]

print(high_marks)
```

The first `mark` is the value we keep.

`for mark in marks` checks each value.

`if mark >= 70` keeps only the values that match the condition.

## Task 12: Filter a list

Use a list comprehension to create a new list called `busy_weeks`.

It should keep only the values from `weekly_hours` that are 5 or more.

Print the new list.

```python
busy_weeks = [hours for hours in weekly_hours if hours >= 5]

print(busy_weeks)
```
