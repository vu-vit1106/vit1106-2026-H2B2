## Why learn NumPy?

Python lists are useful. You already used lists to store several values.

NumPy is useful when the list becomes large and we want to do the same operation to many numbers.

| Python list | NumPy array |
|---|---|
| Good for general values | Good for many numbers |
| Can mix different types | Usually one main data type |
| Often uses loops | Can work on all values at once |
| Built into Python | Imported with `import numpy as np` |

The main idea is not complicated: a NumPy array is a container for values, often numbers.

## Import NumPy

We usually import NumPy as `np`.

This is a common short name used by Python programmers.

````python
import numpy as np

print("NumPy is ready")
print(np.__version__)
````

## A list and an array

The values below are simple weekly study hours.

First, we create a normal Python list.

Then, we create a NumPy array with the same values.

````python
weekly_hours_list = [2, 4, 5, 3, 6]
weekly_hours_array = np.array([2, 4, 5, 3, 6])

print(weekly_hours_list)
print(weekly_hours_array)
````

## They look similar

A list and an array can look similar when printed.

But they do not behave exactly the same.

The next cell shows their types.

````python
print(type(weekly_hours_list))
print(type(weekly_hours_array))
````

## Adding a number

With a Python list, `+ 1` does not add 1 to every value.

The next cell shows the error. The code is inside `try` and `except` so the notebook can keep running.

````python
try:
    print(weekly_hours_list + 1)
except TypeError as error:
    print("Python list error:")
    print(error)
````

## NumPy can work on all values at once

With a NumPy array, `+ 1` adds 1 to every value.

This is one of the main reasons NumPy is useful.

````python
extra_hours_array = weekly_hours_array + 1

print(extra_hours_array)
````

## Data types

A data type describes what kind of values are stored.

Common examples:

| Data type | Meaning | Example |
|---|---|---|
| integer | whole number | `5` |
| float | number with a decimal point | `5.5` |
| string | text | `"hello"` |

NumPy arrays usually work best when the values are the same type.

````python
whole_numbers = np.array([1, 2, 3, 4])
decimal_numbers = np.array([1.0, 2.5, 3.2, 4.8])
words = np.array(["red", "green", "blue"])

print(whole_numbers.dtype)
print(decimal_numbers.dtype)
print(words.dtype)
````

## Lists can mix types

A Python list can mix numbers and text.

This can be useful, but it can also make number work less clear.

````python
mixed_list = [1, "two", 3.0]

print(mixed_list)
print(type(mixed_list[0]))
print(type(mixed_list[1]))
print(type(mixed_list[2]))
````

## NumPy chooses one type when possible

If an array mixes numbers and text, NumPy may convert values so they fit into one array.

This is one reason we usually keep NumPy arrays focused on one kind of data.

````python
mixed_array = np.array([1, "two", 3.0])

print(mixed_array)
print(mixed_array.dtype)
````

## A simple NumPy calculation

NumPy can also do calculations on every value.

The next cell adds 10 to every number in the array.

````python
sample_marks = np.array([45, 60, 72, 88])
bonus_marks = sample_marks + 10

print(sample_marks)
print(bonus_marks)
````

## Useful NumPy functions

NumPy has functions for common summaries.

These examples use marks, but the same idea could work with prices, hours, temperatures, or many other numbers.

````python
marks = np.array([45, 60, 72, 88])

print("Lowest mark:", np.min(marks))
print("Highest mark:", np.max(marks))
print("Average mark:", np.mean(marks))
print("Total marks:", np.sum(marks))
````

## Filtering with NumPy

Filtering means keeping only the values we want.

The next cells keep marks that are 50 or higher:

```python
passing_marks = marks[marks >= 50]
```

This line does two things, from the inside out:

1. `marks >= 50` compares every value in `marks` to `50` and creates a new array of `True` / `False` values — one for each mark. This is called a **boolean mask**.
2. `marks[...]` then uses that mask as an index into `marks`. NumPy keeps only the values lined up with `True`, and drops the rest.

The next cell shows the mask on its own, so you can see the two steps separately before they are combined.

````python
mask = marks >= 50

print(mask)
````

````python
passing_marks = marks[marks >= 50]

print(passing_marks)
````

## Why speed matters

For five numbers, speed does not matter.

For thousands or millions of numbers, speed can matter.

The next examples use a large list of numbers. We will filter the numbers and keep only values above a limit.

This is not advanced maths. We are only asking:

> Is this number bigger than the limit?

````python
MAX_ELEMENTS = 1_000_000
LIMIT = 500_000

numbers_list = list(range(MAX_ELEMENTS))
numbers_array = np.arange(MAX_ELEMENTS)

print("Number of elements:", MAX_ELEMENTS)
print("Limit:", LIMIT)
````

## Method 1: A normal loop

This code uses a `for` loop.

It checks each number one by one and appends values that match the condition.

This is easy to read, but it can be slower for large data.

````python
import time

start_time = time.perf_counter()

filtered_loop = []
for number in numbers_list:
    if number > LIMIT:
        filtered_loop.append(number)

loop_time = time.perf_counter() - start_time

print("Values kept:", len(filtered_loop))
print("Loop time:", round(loop_time, 4), "seconds")
````

## Method 2: A list comprehension

A list comprehension does the same kind of work in a shorter form.

It is still working with a Python list, but it is usually faster than writing the loop yourself.

````python
start_time = time.perf_counter()

filtered_comprehension = [number for number in numbers_list if number > LIMIT]

comprehension_time = time.perf_counter() - start_time

print("Values kept:", len(filtered_comprehension))
print("List comprehension time:", round(comprehension_time, 4), "seconds")
````

## Method 3: NumPy filtering

NumPy can check the whole array in one operation.

The expression `numbers_array > LIMIT` creates many True/False values — the same boolean mask idea you saw earlier in *Filtering with NumPy*.

NumPy then keeps the numbers where the answer is True.

````python
start_time = time.perf_counter()

filtered_numpy = numbers_array[numbers_array > LIMIT]

numpy_time = time.perf_counter() - start_time

print("Values kept:", len(filtered_numpy))
print("NumPy time:", round(numpy_time, 4), "seconds")
````

## Compare the times

The exact times will be different on different computers.

The pattern is the important part:

- a normal loop is often the slowest
- a list comprehension is often faster
- NumPy is often faster again for large numeric data

Small differences do not matter here. We are looking for the general idea.

````python
print("Loop:", round(loop_time, 4), "seconds")
print("List comprehension:", round(comprehension_time, 4), "seconds")
print("NumPy:", round(numpy_time, 4), "seconds")

print()
print("Speed compared with the loop:")
print("Comprehension is about", round(loop_time / comprehension_time, 2), "times faster than the loop")
print("NumPy is about", round(loop_time / numpy_time, 2), "times faster than the loop")
````
