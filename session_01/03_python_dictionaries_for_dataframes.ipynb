## Why dictionaries matter

Lists store values by position.

Dictionaries store values by name.

That small difference is important for data work. A table column has a name, such as `student`, `campus`, or `score`. Dictionaries help us practise using names to organise data before we work with full DataFrames.

## Creating a dictionary

A dictionary uses key-value pairs.

- The key is the label.
- The value is the data stored under that label.

The keys below are `name`, `campus`, and `score`.

```python
student = {
    "name": "Taylor Nguyen",
    "campus": "Footscray",
    "score": 78
}

print(student)
```

## Reading values from a dictionary

Use square brackets and the key name to read one value.

This is similar to asking: what is the value stored under this label?

```python
print(student["name"])
print(student["score"])
```

## Adding and updating values

Dictionaries can change.

If the key is new, Python adds it.

If the key already exists, Python updates the value.

```python
student["course"] = "VIT1106"
student["score"] = 82

print(student)
```

## Safer reading with `.get()`

If you use square brackets with a key that does not exist, Python gives an error.

The `.get()` method lets you provide a default value instead.

```python
print(student.get("campus"))
print(student.get("attendance", "not recorded"))
```

## Looking at keys, values, and items

A dictionary gives us three common views of the data:

- `.keys()` shows the labels
- `.values()` shows the stored values
- `.items()` shows key-value pairs

The `.items()` view is useful when we want to loop through a dictionary.

```python
print(student.keys())
print(student.values())
print(student.items())
```

## Looping through a dictionary

A loop can process each key-value pair.

This is useful for checking, printing, or preparing data.

```python
for key, value in student.items():
    print(f"{key}: {value}")
```

## A list of dictionaries

One dictionary can store one record.

A list of dictionaries can store many records.

This is a common pattern before working with data tables.

```python
students = [
    {"name": "Taylor Nguyen", "campus": "Footscray", "score": 82},
    {"name": "Mia Chen", "campus": "City", "score": 74},
    {"name": "Jordan Lee", "campus": "Footscray", "score": 46}
]

print(students[0])
print(students[0]["name"])
```

## Looping through records

When we have a list of dictionaries, a loop lets us process one record at a time.

```python
for record in students:
    print(student_summary(record))
```

## Dictionary of lists

A table can also be represented as a dictionary of lists.

Each key can act like a column name.

Each list can act like the values in that column.

```python
student_table = {
    "name": ["Taylor Nguyen", "Mia Chen", "Jordan Lee"],
    "campus": ["Footscray", "City", "Footscray"],
    "score": [82, 74, 46]
}

print(student_table["name"])
print(student_table["score"])
```
