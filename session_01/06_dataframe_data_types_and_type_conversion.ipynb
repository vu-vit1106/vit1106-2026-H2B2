# DataFrame data types and type conversion

This notebook shows how to inspect column data types in a pandas DataFrame and how to change them when the stored type is not suitable for analysis.

You have already seen DataFrames as tables. In this notebook, the focus is on the **type of each column**.

By the end of this notebook, you should be able to:

- display the data types in a DataFrame
- explain the difference between a stored type and an analysis type
- convert columns to numeric, category, string, and date types
- check that a conversion worked
- avoid changing a column just because it contains digits

## Why data types matter

A DataFrame stores each column with a data type. The data type affects what pandas can do with the column.

For example:

- numbers can be averaged
- text can be searched or combined
- dates can be sorted by time
- categories can be counted by group

A column may look correct but still be stored in the wrong way. We inspect before we analyse.

## Import pandas

```python
import pandas as pd

print("pandas is ready")
```

## Create a small DataFrame

The next table is deliberately small so you can see every value.

Some columns are stored as text even though we may want to use them as numbers or dates later.

```python
df_students = pd.DataFrame({
    "student": ["Taylor Nguyen", "Mia Chen", "Jordan Lee", "Priya Shah"],
    "campus": ["Footscray", "City", "Footscray", "Sunshine"],
    "postcode": ["3011", "3000", "3011", "3020"],
    "study_hours": ["6", "5", "3", "7"],
    "attendance": ["90", "86", "72", "95"],
    "score": [82, 74, 46, 88],
    "submitted": ["2026-03-05", "2026-03-04", "2026-03-07", "2026-03-03"],
    "support_level": ["low", "medium", "high", "low"],
})

df_students
```

## Step 1: inspect the data types

Use `.dtypes` to display the stored type of each column.

```python
df_students.dtypes
```

You will see names such as `object` and `int64`.

For now:

- `object` usually means text or mixed values
- `int64` means whole numbers
- `float64` means decimal numbers
- `datetime64` means dates and times
- `category` means a fixed set of labels

## Step 2: compare stored type with analysis type

The stored type is what pandas currently sees. The analysis type is how we should treat the variable.

| column | current stored type | how we should analyse it |
|---|---|---|
| student | object | text label |
| campus | object | category |
| postcode | object | category, not a number |
| study_hours | object | number |
| attendance | object | number |
| score | int64 | number |
| submitted | object | date |
| support_level | object | ordered category |

This table is the plan for the conversions below.

## Step 3: convert text numbers to numeric columns

`study_hours` and `attendance` are stored as text. They should be numbers, so we convert them with `pd.to_numeric()`.

```python
df_students["study_hours"] = pd.to_numeric(df_students["study_hours"])
df_students["attendance"] = pd.to_numeric(df_students["attendance"])

df_students.dtypes
```

Now pandas can calculate numeric summaries.

```python
df_students[["study_hours", "attendance", "score"]].mean()
```

## Step 4: convert labels to categories

`campus` is a label chosen from a small set of values. It is useful to store it as a category.

```python
df_students["campus"] = df_students["campus"].astype("category")

df_students.dtypes
```

A category column is useful for counting groups.

```python
df_students["campus"].value_counts()
```

## Step 5: keep postcode as a category

A postcode contains digits, but it is not a measured quantity. We should not average it.

Here we convert it to a category, not to a number.

```python
df_students["postcode"] = df_students["postcode"].astype("category")

df_students.dtypes
```

## Step 6: convert text dates to real dates

The `submitted` column is stored as text. Convert it with `pd.to_datetime()` so pandas understands it as dates.

```python
df_students["submitted"] = pd.to_datetime(df_students["submitted"])

df_students.dtypes
```

Now date operations work properly.

```python
df_students.sort_values("submitted")
```

## Step 7: convert an ordered category

`support_level` has an order: low, medium, high. We can tell pandas the order explicitly.

```python
support_order = pd.CategoricalDtype(
    categories=["low", "medium", "high"],
    ordered=True,
)

df_students["support_level"] = df_students["support_level"].astype(support_order)

df_students.dtypes
```

Now pandas knows the category order.

```python
df_students.sort_values("support_level")
```

Always check the result after changing data types.
