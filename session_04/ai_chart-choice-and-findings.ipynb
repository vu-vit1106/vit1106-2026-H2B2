## Step 4: build the three charts

Build the chart you judged correct for each question — whether that was your
first choice or AI's.

```python
sns.histplot(data=races, x="mark_seconds")
plt.title("Distribution of race times")
plt.show()
```

```python
top_countries = races["country"].value_counts().head(5).index

sns.boxplot(
    data=races[races["country"].isin(top_countries)],
    x="country",
    y="mark_seconds",
)
plt.title("Race times for the top five countries")
plt.show()
```

```python
performances_by_decade = races["decade"].value_counts().sort_index()

sns.barplot(
    x=performances_by_decade.index.astype(str),
    y=performances_by_decade.values,
)
plt.title("Top performances by decade")
plt.xlabel("decade")
plt.ylabel("performances")
plt.show()
```

## Step 5: build the evidence table

Make a crosstab of country against decade for the top five countries, with row
percentages, and keep it visible — it is the evidence you will check AI's
findings against.

```python
subset = races[races["country"].isin(top_countries)]

counts = pd.crosstab(subset["country"], subset["decade"], margins=True)
row_pct = pd.crosstab(
    subset["country"], subset["decade"], normalize="index"
).round(2)

print(counts)
row_pct
```
