## Session 4 · Explore — Frequencies and Crosstabs (Python)

**Dataset:** `w1500_clean.csv` — your cleaned 1500m data from Notebook 1.

Categories are summarised by **counting**. This notebook covers the three pandas
tools: `value_counts()` (one category), `crosstab()` (two categories), and
`groupby()` (the general tool). Replace every `# TODO` and run the cell.

### 1. Load the cleaned data

Read `data/w1500_clean.csv` into `df` and print its shape.

```python
import pandas as pd
from pathlib import Path

DATA_FOLDER = Path("../data")
DATA_FILE = DATA_FOLDER / "w1500_clean.csv"

df = pd.read_csv(DATA_FILE)
print(df.shape)
```

### 2. Count one category

Use `value_counts()` on `country`. Which countries have the most performances?

```python
df["country"].value_counts().head(10)
```

### 3. Relative frequency (percentages)

Add `normalize=True` to get proportions, multiply by 100 for percentages, and round to 1 decimal.

```python
(df["country"].value_counts(normalize=True) * 100).round(1).head(10)
```

### 4. Cross-tabulate two categories

`pd.crosstab(rows, cols)` counts combinations. Cross-tab `country` (rows) against `decade` (cols), with `margins=True` for totals. Show a few leading countries.

```python
ct = pd.crosstab(df["country"], df["decade"], margins=True)
ct.sort_values("All", ascending=False).head(8)
```

### 5. Row percentages — a fair comparison

Raw counts mislead when totals differ. Turn each **row** into percentages with `normalize="index"` so every country sums to 100%. Now you compare *when* each country peaked.

```python
(pd.crosstab(df["country"], df["decade"], normalize="index") * 100).round(1).head(8)
```

### 6. The same thing with `groupby()`

`value_counts()` is a shortcut; `groupby()` is the general tool. Reproduce the per-country count with `df.groupby("country").size()` and confirm it matches step 2.

```python
df.groupby("country").size().sort_values(ascending=False).head(10)
```
