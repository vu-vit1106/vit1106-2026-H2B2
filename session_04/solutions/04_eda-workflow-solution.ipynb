### 1. Structure — what have I got?

Read the cleaned data, then print `shape`, `dtypes`, and the first rows.

```python
import pandas as pd
from pathlib import Path

DATA_FOLDER = Path("../data")
DATA_FILE = DATA_FOLDER / "w1500_clean.csv"

df = pd.read_csv(DATA_FILE)
print(df.shape)
print(df.dtypes)
df.head()
```

### 2. Data quality — anything missing?

Print `df.isna().sum()`. How many unique athletes are there compared with the number of rows?

```python
print(df.isna().sum())
print("rows:", df.shape[0], "unique athletes:", df["athlete"].nunique())
```

### 3. Summarise one category and one number

`value_counts()` on `country` (top 8), and `describe()` on `mark_seconds`.

```python
print(df["country"].value_counts().head(8))
print(df["mark_seconds"].describe().round(2))
```

### 4. See the shape

Set up seaborn and draw a histogram of `mark_seconds`. Describe the shape in one sentence.

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.set_theme(style="whitegrid")
```

```python
sns.histplot(data=df, x="mark_seconds", bins=25)
plt.title("1500m times")
plt.show()
```

### 5. Look for a relationship

Do older athletes run slower? Make a scatter plot of `age_at_race` (x) against `mark_seconds` (y). This is a **two-variable** question — the kind Session 5 picks up.

```python
sns.scatterplot(data=df, x="age_at_race", y="mark_seconds", alpha=0.3)
plt.title("Age vs race time")
plt.show()
```

### 6. A grouped comparison

Finish with a ranked summary: mean `mark_seconds` per `decade`, sorted by decade.

```python
df.groupby("decade")["mark_seconds"].mean().round(2).sort_index()
```
