## Session 4 · Explore/Visualise — Distribution Plots (Python)

**Dataset:** `w1500_clean.csv`.

Two charts show the **shape** of a numeric variable: a **histogram** (shape in
detail) and a **box plot** (five-number summary + outliers, great for comparing
groups). We build both on `mark_seconds` with seaborn. Replace every `# TODO`.

### 1. Load the data and set up plotting

```python
import pandas as pd
from pathlib import Path

DATA_FOLDER = Path("../data")
DATA_FILE = DATA_FOLDER / "w1500_clean.csv"

df = pd.read_csv(DATA_FILE)
print(df.shape)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")
```

### 2. Histogram — the shape of the times

Use `sns.histplot(data=df, x="mark_seconds", bins=25)` and give it a title. Where is the peak? Is it symmetric, or is there a tail?

```python
sns.histplot(data=df, x="mark_seconds", bins=25)
plt.title("Distribution of all-time women's 1500m times")
plt.xlabel("Race time (seconds)")
plt.show()
```

*Your interpretation:* describe the shape in one sentence. (Hint: remember this is a filtered top-list.)

### 3. Box plot — the five-number summary at a glance

Use `sns.boxplot(data=df, x="mark_seconds")`. The box spans Q1–Q3, the line is the median, whiskers reach the extremes, dots are outliers. Add a title.

```python
sns.boxplot(data=df, x="mark_seconds")
plt.title("1500m times — box plot")
plt.xlabel("Race time (seconds)")
plt.show()
```

### 4. Compare groups with side-by-side box plots

A box plot's power is comparison. Draw `mark_seconds` split by country, for the six countries with the most performances.

```python
top6 = df["country"].value_counts().head(6).index
subset = df[df["country"].isin(top6)]
sns.boxplot(data=subset, x="country", y="mark_seconds")
plt.title("1500m times by leading nation")
plt.show()
```

### 5. Split a histogram by a group (`hue`)

Redraw the histogram for the top 3 countries with `hue="country"` and `element="step"` to compare them on the same axes.

```python
top3 = df["country"].value_counts().head(3).index
sub3 = df[df["country"].isin(top3)]
sns.histplot(data=sub3, x="mark_seconds", hue="country", element="step")
plt.title("1500m times — top 3 nations compared")
plt.show()
```
