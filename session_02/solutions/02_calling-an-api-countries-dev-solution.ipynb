**API:** [countries.dev](https://countries.dev) — a free, public REST API for
country data. No API key, no sign-up, no rate limit: you send a request and get
JSON back.

For every country it can return ISO codes, capital, region and subregion,
population, area, currencies, languages, calling codes, flags, and more. Today
you only need a handful of these fields.

In this notebook you will:

1. Call the API's `/countries` endpoint with `requests`.
2. Turn the JSON response into a pandas DataFrame.
3. Take a first look at what came back.

### 1. Call the API

The endpoint is `GET https://countries.dev/countries`. On its own it returns
every field for every country; the `fields` query parameter lets you ask for
only the columns you want — a comma-separated list.

Request just these fields: `name, capital, region, subregion, population, area,
alpha2Code, alpha3Code`.

Build the URL, call it with `requests.get()`, and check `response.status_code`
— `200` means the request succeeded.

```python
import requests
import pandas as pd
from pathlib import Path

FIELDS = "name,capital,region,subregion,population,area,alpha2Code,alpha3Code"
URL = f"https://countries.dev/countries?fields={FIELDS}"

response = requests.get(URL)
print(response.status_code)
```

### 2. Turn the JSON into a DataFrame

A successful response's body is JSON. Call `response.json()` to turn it into a
Python list of dictionaries — one dictionary per country. Since every
dictionary already has the same flat set of keys, `pd.DataFrame(...)` can build
the table directly, no extra reshaping needed.

Print how many countries came back.

```python
data = response.json()
df_countries = pd.DataFrame(data)
print(len(df_countries), "countries")
df_countries.head()
```

### 3. First look at the data

Before using any dataset, check its shape and types.

```python
print(df_countries.shape)
print(df_countries.dtypes)
print(df_countries.isna().sum())
```

### 4. Save to Excel

Write `countries` to `data/countries.xlsx` with `to_excel()`. Pass
`index=False` so pandas doesn't add its row-number index as an extra column,
and `sheet_name="Countries"`.

```python
DATA_FOLDER = Path("../data")
OUT_FILE = DATA_FOLDER / "countries.xlsx"
df_countries.to_excel(OUT_FILE, sheet_name="Countries", index=False)
print("Wrote:", OUT_FILE.resolve())
```
