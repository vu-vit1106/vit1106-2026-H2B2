## Step 1: import the libraries


```python
from pathlib import Path

import pandas as pd
import requests
from bs4 import BeautifulSoup
```

## Step 2: choose your page

Replace the URL with the page approved by your teacher.


```python
url = "PASTE_APPROVED_URL_HERE"
headers = {
    "User-Agent": "VIT1106 student practice (one polite request)"
}

print(url)
```

## Step 3: fetch the page

Run this cell only after your teacher has approved the page.


```python
response = requests.get(url, headers=headers, timeout=30)
print("Status code:", response.status_code)
print("Downloaded characters:", len(response.text))
```

## Step 4: create the BeautifulSoup object


```python
soup = BeautifulSoup(response.text, "html.parser")

page_title = soup.find("title")
if page_title is not None:
    print(page_title.get_text(" ", strip=True))
```

## Step 5: inspect possible tags

Start with common tags. This helps you decide what is easy and sensible to extract.


```python
for tag_name in ["h1", "h2", "h3", "a", "li"]:
    tags = soup.find_all(tag_name)
    print(tag_name, "tags found:", len(tags))

    for tag in tags[:5]:
        text = tag.get_text(" ", strip=True)
        if text:
            print(" -", text[:100])

    print()
```

## Step 6: extract one set of information

Adapt this example. It extracts link text and link URLs from `<a>` tags.

If your page is better suited to headings, list items, or table rows, ask AI to help you adapt the loop while keeping the code simple.


```python
rows = []

for tag in soup.find_all("a"):
    text = tag.get_text(" ", strip=True)
    href = tag.get("href")

    if text and href:
        rows.append({
            "text": text,
            "href": href,
        })

df_result = pd.DataFrame(rows)
print("Rows extracted:", len(df_result))
df_result.head(10)
```

## Step 7: save your CSV file

Change the filename so it describes your page and extraction.


```python
DATA_FOLDER = Path("data")
DATA_FOLDER.mkdir(exist_ok=True)

OUTPUT_FILE = DATA_FOLDER / "student_web_scraping_result.csv"
df_result.to_csv(OUTPUT_FILE, index=False)

print("Saved", len(df_result), "rows to", OUTPUT_FILE)
```
