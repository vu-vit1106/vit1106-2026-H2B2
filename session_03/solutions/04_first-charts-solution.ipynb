## Session 3 · Notebook 4 — First Charts with seaborn (Python)

**Dataset:** `Y1_T1_2025.xlsx` — Term 1, 2025 first-year unit-attempt extract.

Numbers describe; **pictures reveal**. This notebook builds the three workhorse
charts on the real data using **seaborn**:

- **Bar chart** — compare a value across categories
- **Histogram** — see the shape of one numeric variable
- **Scatter plot** — see the relationship between two numeric variables

This solution copy shows one possible completed version.

### 1. Load the data

Read the **`Extract`** sheet into `df` and print its shape.

```python
import pandas as pd
import numpy as np
from pathlib import Path

DATA_FOLDER = Path("../data")
DATA_FILE = DATA_FOLDER / "Y1_T1_2025.xlsx"

df = pd.read_excel(DATA_FILE, sheet_name="Extract")
print(df.shape)
```

### The plotting stack — where seaborn fits

- **matplotlib** is the low-level engine that draws the pixels. Powerful, but
  verbose — a simple chart can take many lines.
- **seaborn** sits on top of matplotlib. One call (`sns.histplot(...)`,
  `sns.barplot(...)`) produces a good-looking chart and does the tedious work for
  you.

We use **seaborn** to get to insight faster, and reach *down* into matplotlib
(`plt.title(...)`, `plt.xlabel(...)`, `plt.show()`) only to add titles, labels or
to save a figure. Run the setup cell once.

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")
```

### 2. Bar chart — a value across categories

Use `sns.barplot(data=df, x=..., y=...)` to show the **mean `mark` for each
`program_code`**. seaborn computes the mean per category automatically (the thin
line on each bar is a 95% confidence interval). Add a title and a y-axis label,
then `plt.show()`.

```python
sns.barplot(data=df, x="program_code", y="mark")
plt.title("Mean mark by program")
plt.ylabel("mean mark")
plt.show()
```

### 3. Histogram — the shape of one variable

Use `sns.histplot(data=df, x="mark", bins=20)` to show the distribution of marks.
Give it a title. Where is the peak — is it symmetric, or are there tails?

```python
sns.histplot(data=df, x="mark", bins=20)
plt.title("Distribution of marks")
plt.show()
```

### 4. Colour by a category (`hue`)

seaborn splits a chart by a category with one argument: `hue`. Redraw the
histogram with `hue="international_flag"` and `element="step"` to compare domestic
vs international students on the same axes.

```python
sns.histplot(data=df, x="mark", bins=20, hue="international_flag", element="step")
plt.title("Marks: domestic vs international")
plt.show()
```
