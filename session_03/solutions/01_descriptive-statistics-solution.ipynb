### 3. Measures of centre

Compute, for the `mark` column:

- the **mean** with `.mean()`
- the **median** with `.median()`
- the **mode** with `.mode()`

Print all three.

> `.mode()` returns a *Series* (there can be more than one most-common value), so
> read it with `.mode().tolist()` or `[0]`.

```python
print("mean  :", round(df["mark"].mean(), 2))
print("median:", df["mark"].median())
print("mode  :", df["mark"].mode().tolist())
```

*Your interpretation:* is the mean above or below the median — is the
distribution roughly symmetric, or pulled to one side?

### 4. Mode of a *categorical* column

The mode is the only measure of centre that also works on text. Find the single
most common **`grade_code`** using `.mode()`, and confirm it with
`value_counts().head()`.

```python
print("most common grade:", df["grade_code"].mode().tolist())
df["grade_code"].value_counts().head()
```

### 5. Measures of variation

Now describe how *spread out* the marks are:

- **Range** = maximum − minimum (`.max()` and `.min()`)
- **Standard deviation** with `.std()`

Print the min, max, range and standard deviation.

> By default pandas uses the **sample** standard deviation (`ddof=1`). That's the
> right default here; you don't need to change it.

```python
lo, hi = df["mark"].min(), df["mark"].max()
print("min:", lo, " max:", hi, " range:", hi - lo)
print("std dev:", round(df["mark"].std(), 2))
```
