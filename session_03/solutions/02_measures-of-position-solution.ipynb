## Session 3 · Notebook 2 — Measures of Position (Python)

**Dataset:** `Y1_T1_2025.xlsx` — Term 1, 2025 first-year unit-attempt extract.

Centre and spread describe the whole column. **Position** asks where *one* value
sits compared to the rest. This notebook works the three tools from the theory on
the `mark` column:

- **percentiles and quartiles** with `.quantile()`
- the **IQR** and the 1.5 × IQR **outlier** rule
- **z-scores** — standardising a value as "how many standard deviations from the
  mean"

This solution copy shows one possible completed version.

### 1. Load the data

Read the **`Extract`** sheet into `df` and print its shape.

```python
import pandas as pd
import numpy as np
from pathlib import Path

DATA_FOLDER = Path("../data")
DATA_FILE = DATA_FOLDER / "Y1_T1_2025.xlsx"

df = pd.read_excel(DATA_FILE, sheet_name="Extract")
print(df.shape)
```

### 2. Percentiles

`.quantile(p)` returns the value at proportion `p` (so `0.90` is the 90th
percentile). For the `mark` column, print the 25th, 50th and 90th percentiles.

> The 90th percentile is the mark you'd need to beat 90% of attempts.

```python
print("25th:", df["mark"].quantile(0.25))
print("50th:", df["mark"].quantile(0.50))
print("90th:", df["mark"].quantile(0.90))
```

### 3. Quartiles

The quartiles are just three percentiles: **Q1** = 0.25, **Q2** = 0.50 (the
median), **Q3** = 0.75. Store them in `q1`, `q2`, `q3` and print them.

```python
q1 = df["mark"].quantile(0.25)
q2 = df["mark"].quantile(0.50)
q3 = df["mark"].quantile(0.75)
print("Q1:", q1, " Q2 (median):", q2, " Q3:", q3)
```

### 4. The IQR and outlier fences

The **IQR** is the spread of the middle half: `IQR = Q3 - Q1`. The 1.5 × IQR rule
flags possible outliers outside these fences:

- lower fence = `Q1 - 1.5 * IQR`
- upper fence = `Q3 + 1.5 * IQR`

Compute the IQR and both fences, then count how many marks fall outside them.

```python
iqr = q3 - q1
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr
print("IQR:", iqr, " lower fence:", lower, " upper fence:", upper)

df_outliers = df[(df["mark"] < lower) | (df["mark"] > upper)]
print("outliers flagged:", len(df_outliers))
```

*Your interpretation:* did the rule flag any marks? Given marks are capped at

### 5. Z-scores — standardising a value

A **z-score** is `(value - mean) / std` — how many standard deviations a value
sits from the mean. Add a `mark_z` column that standardises every mark, then show
a few rows of `mark` next to `mark_z`.

> Missing marks stay missing (`NaN`) — that's fine.

```python
mean = df["mark"].mean()
std = df["mark"].std()
df["mark_z"] = (df["mark"] - mean) / std
df[["mark", "mark_z"]].head()
```

### 6. Use the z-score to flag unusual marks

A common rule: a z-score beyond **−3 or +3** is unusual. Count how many marks have
`abs(mark_z) > 3`, and show a few of them.

```python
df_unusual = df[df["mark_z"].abs() > 3]
print("unusual marks (|z| > 3):", len(df_unusual))
df_unusual[["mark", "mark_z"]].head()
```
