### The plotting stack — where seaborn fits

- **matplotlib** is the low-level engine that draws the pixels. Powerful, but
  verbose — a simple chart can take many lines.
- **seaborn** sits on top of matplotlib. One call (`sns.histplot(...)`,
  `sns.barplot(...)`) produces a good-looking chart and does the tedious work for
  you.

We use **seaborn** to get to insight faster, and reach *down* into matplotlib
(`plt.title(...)`, `plt.xlabel(...)`, `plt.show()`) only to add titles, labels or
to save a figure. Run the setup cell once.

````python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")
````
