## 1. Start with one dataset

Run the setup cell. It imports the libraries, records where the two data files are,
and loads the student habits data into `habits`.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

sns.set_theme(style="whitegrid", palette="colorblind")

DATA_FOLDER = Path("../data")
HABITS_FILE = DATA_FOLDER / "student_habits.csv"
SALARY_FILE = DATA_FOLDER / "experience_and_salary.csv"
ANSCOMBE_FILE = DATA_FOLDER / "anscombe.csv"

habits = pd.read_csv(HABITS_FILE)
habits.head()
```

### Activity: Read the columns

Look at the first five rows. Which column belongs on the horizontal axis and which
belongs on the vertical axis for this question?

> Do students who report more study hours also tend to have higher exam scores?

**Model answer:** Put `study_hours` on the horizontal axis and `exam_score` on the vertical axis.

## 2. See the relationship with a scatter plot

Create the scatter plot shown in Topic 01. Each dot should represent one student.

```python
sns.scatterplot(
    data=habits,
    x="study_hours",
    y="exam_score",
)
plt.title("Study hours and exam score")
plt.xlabel("Study hours")
plt.ylabel("Exam score")
plt.show()
```

### Activity: Describe before calculating

Describe the scatter plot in one sentence. Include its direction and whether the
dots look tightly or loosely grouped around a straight line.

**Model answer:** The plot shows a strong positive relationship because the dots rise from left to right and stay fairly close to a straight-line pattern.

## 3. Measure direction and strength

Calculate r for `study_hours` and `exam_score`. Store the result so it can be used
again, then print it rounded to two decimal places.

```python
study_exam_r = habits["study_hours"].corr(habits["exam_score"])
print("Study hours and exam score:", round(study_exam_r, 2))
```

**Model answer:** r is approximately +0.92, so study hours and exam score have a strong positive linear relationship.

## 4. Compare all pairs in the student habits data

Create the correlation matrix for the numeric columns in `habits`. Round every
value to two decimal places and store the result as `habit_correlations`.

```python
habit_correlations = habits.corr(numeric_only=True).round(2)
habit_correlations
```

### Display the matrix as a heatmap

Use `habit_correlations` as the heatmap data. Add the correlation values to the
cells and keep the colour scale fixed from −1 to +1.

```python
sns.heatmap(
    habit_correlations,
    annot=True,
    cmap="RdBu_r",
    vmin=-1,
    vmax=1,
)
plt.title("Correlations in the student habits data")
plt.show()
```

### Activity: Investigate one more pair

Choose one pair from Topic 01:

- attendance and exam score, or
- screen time and sleep hours

Find its heatmap value, state its direction and strength, then create the matching
scatter plot.

```python
x_column = "screen_time"
y_column = "sleep_hours"

selected_r = habits[x_column].corr(habits[y_column])
print("Selected correlation:", round(selected_r, 2))

sns.scatterplot(
    data=habits,
    x=x_column,
    y=y_column,
)
plt.title("Screen time and sleep hours")
plt.xlabel("Screen time (hours)")
plt.ylabel("Sleep (hours)")
plt.show()
```

**Model answer:** Screen time and sleep hours have r approximately −0.87. This is a strong negative linear relationship, and the scatter plot falls from left to right.

## 6. Always plot — a number can hide the shape

Anscombe's quartet contains four datasets with approximately the same correlation.
The file location and loading code are supplied. You will calculate each correlation
and create the four plots.

```python
anscombe = pd.read_csv(ANSCOMBE_FILE)
anscombe.head()
```

```python
for dataset_name, group in anscombe.groupby("dataset"):
    dataset_r = group["x"].corr(group["y"])
    print(dataset_name, round(dataset_r, 2))

sns.lmplot(
    data=anscombe,
    x="x",
    y="y",
    col="dataset",
    col_wrap=2,
    ci=None,
)
plt.show()
```

### Activity: Trust the number?

Imagine that two variables have r = 0.0. Describe two possible scatter plots:

1. one where there is no clear relationship
2. one where there is a strong curved relationship

**Model answer:** The first plot could be a shapeless cloud with no visible pattern. The second could be a clear U-shaped curve whose downward and upward sections cancel when summarised by one straight-line correlation.

### Activity: Apply `lmplot` to student habits

Use `sns.lmplot()` with the `habits` dataframe. Put `study_hours` on the horizontal
axis and `exam_score` on the vertical axis. Hide the confidence interval.

Compare the result with your earlier scatter plot. What does the fitted line add,
and what detail remains visible only in the dots?

```python
sns.lmplot(
    data=habits,
    x="study_hours",
    y="exam_score",
    ci=None,
)
plt.title("Study hours and exam score with a fitted line")
plt.xlabel("Study hours")
plt.ylabel("Exam score")
plt.show()
```

**Model answer:** The fitted line makes the overall positive linear trend easier to see. The individual dots still show the spread of student results, repeated values and students whose scores sit away from the line.
