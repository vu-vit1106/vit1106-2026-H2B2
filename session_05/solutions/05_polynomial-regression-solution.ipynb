## Investigation question

> **How is a runner's age related to their 1500 m race time?**

Decide whether a straight line is suitable before fitting it. Then compare it with a quadratic curve.

## 1. Predict before analysing

Age is measured in years. Race time is measured in seconds, where lower is faster.

1. Do you expect race time to continually rise or continually fall from age 15 to 45?
2. Would one straight line represent the complete relationship?
3. Sketch the relationship you expect and explain why.

**Answer:** A line is a useful starting point, but I expect times to fall, reach a low point, and later rise. One straight line is unlikely to represent the complete relationship.

## 2. Load the prepared data

This supplied setup loads the data and prepares a reproducible display sample. All models use the complete dataset.

```python
from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

DATA_FOLDER = Path("../data" if Path.cwd().name == "solutions" else "data")
RACE_FILE = DATA_FOLDER / "w1500_age_time_2001_2025.csv"

races = pd.read_csv(RACE_FILE)
age = races["age_at_race"].to_numpy()
race_time = races["mark_seconds"].to_numpy()
display_data = races.sample(n=4000, random_state=1106)
age_grid = np.linspace(15, 45, 300)
print(f"Rows: {len(races):,}")
races.head()
```

## 3. Fit the familiar straight line

Fit a degree-1 model. Calculate its predictions and R², then print the slope, intercept, and R².

```python
linear_coefficients = np.polyfit(age, race_time, 1)
linear_slope, linear_intercept = linear_coefficients
linear_predictions = np.polyval(linear_coefficients, age)
total_variation = ((race_time - race_time.mean()) ** 2).sum()
linear_error = ((race_time - linear_predictions) ** 2).sum()
linear_r_squared = 1 - linear_error / total_variation
print(f"Slope: {linear_slope:.3f}")
print(f"Intercept: {linear_intercept:.2f}")
print(f"Linear R-squared: {linear_r_squared:.3f}")
```

### Reproduce the straight-line graph

Plot the sampled observations and place the fitted line over them.

```python
plt.figure(figsize=(9, 5.5))
plt.scatter(display_data["age_at_race"], display_data["mark_seconds"], s=10, alpha=0.12)
plt.plot(age_grid, np.polyval(linear_coefficients, age_grid), linewidth=3)
plt.xlabel("Age at race (years)")
plt.ylabel("1500 m time (seconds)")
plt.title(f"Straight-line fit (R² = {linear_r_squared:.3f})")
plt.show()
```

**Answer:** The negative slope means fitted time falls by about 0.8 seconds per year. R² ≈ 0.112 means the line explains only about 11% of race-time variation, so age alone is weak for individual prediction.

## 4. Compare the line with domain knowledge

A line says times continue falling with age. Sport knowledge suggests improve, reach a fastest period, then decline. Because lower times are better, sketch the resulting expected shape before continuing.

## 5. Fit a quadratic model

Use degree 2 to add one bend. Calculate its R² and fitted minimum age.

```python
quadratic_coefficients = np.polyfit(age, race_time, 2)
quadratic_predictions = np.polyval(quadratic_coefficients, age)
quadratic_error = ((race_time - quadratic_predictions) ** 2).sum()
quadratic_r_squared = 1 - quadratic_error / total_variation
quadratic_a, quadratic_b, quadratic_c = quadratic_coefficients
minimum_age = -quadratic_b / (2 * quadratic_a)
print(f"Quadratic R-squared: {quadratic_r_squared:.3f}")
print(f"Fitted minimum age: {minimum_age:.1f} years")
```

### Plot and compare both models

Show the observations, dashed line, quadratic curve, and fitted minimum.

```python
plt.figure(figsize=(9, 5.5))
plt.scatter(display_data["age_at_race"], display_data["mark_seconds"], s=10, alpha=0.10)
plt.plot(age_grid, np.polyval(linear_coefficients, age_grid), "--", linewidth=2, label=f"Linear: R² = {linear_r_squared:.3f}")
plt.plot(age_grid, np.polyval(quadratic_coefficients, age_grid), linewidth=3, label=f"Quadratic: R² = {quadratic_r_squared:.3f}")
plt.axvline(minimum_age, color="black", linestyle=":", label=f"Minimum ≈ {minimum_age:.1f} years")
plt.xlabel("Age at race (years)")
plt.ylabel("1500 m time (seconds)")
plt.title("Linear and quadratic fits")
plt.legend()
plt.show()
```

**Answer:** Quadratic R² ≈ 0.176 improves on 0.112 and its U-shape is more natural, but it still explains only about 18% of variation. The minimum near 29.3 describes this dataset; it is not a universal best age.
