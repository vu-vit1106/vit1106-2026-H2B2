## 1. Load and refit the Topic 02 models

Run this supplied setup cell before starting the prediction activities.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

sns.set_theme(style="whitegrid", palette="colorblind")

DATA_FOLDER = Path("../data")
HABITS_FILE = DATA_FOLDER / "student_habits.csv"
SALARY_FILE = DATA_FOLDER / "experience_and_salary.csv"

habits = pd.read_csv(HABITS_FILE)
salary = pd.read_csv(SALARY_FILE)

study_slope, study_intercept = np.polyfit(
    habits["study_hours"], habits["exam_score"], 1
)
salary_slope, salary_intercept = np.polyfit(
    salary["years_experience"], salary["salary_aud"], 1
)

print("Study model:", round(study_intercept, 2), round(study_slope, 2))
print("Salary model:", round(salary_intercept, 2), round(salary_slope, 2))
```

## 2. Predict an exam score

Substitute 6 study hours into the fitted model. Store the result as
`predicted_score`.

```python
prediction_hours = 6
predicted_score = study_intercept + study_slope * prediction_hours
print("Predicted score:", round(predicted_score, 1))
```

### Reproduce the study prediction graph

Plot the original student data and regression line. Highlight the prediction at 6
hours and add dashed guides from the point towards both axes. Display the line
equation inside the graph.

```python
study_x = np.linspace(
    habits["study_hours"].min(),
    habits["study_hours"].max(),
    100,
)

plt.scatter(habits["study_hours"], habits["exam_score"], label="Student data")
plt.plot(
    study_x,
    study_intercept + study_slope * study_x,
    label="Regression line",
)
plt.scatter([prediction_hours], [predicted_score])
plt.plot(
    [prediction_hours, prediction_hours],
    [habits["exam_score"].min(), predicted_score],
    "--",
)
plt.plot(
    [habits["study_hours"].min(), prediction_hours],
    [predicted_score, predicted_score],
    "--",
)
plt.text(
    0.04,
    0.95,
    "score = 54.6 + 4.75 × study hours",
    transform=plt.gca().transAxes,
    va="top",
)
plt.xlabel("Study hours")
plt.ylabel("Exam score")
plt.legend()
plt.show()
```

### Activity: Make three predictions

Predict the exam score at 3, 8 and 10 study hours. Print each result.

```python
for hours in [3, 8, 10]:
    score = study_intercept + study_slope * hours
    print(hours, "hours:", round(score, 1))
```

**Model answer:** The 10-hour prediction needs the most caution. It is just beyond the observed maximum of 9.9 hours and predicts slightly above the maximum possible score of 100.

## 3. Predict a salary

Substitute 8 years into the salary model and store the result as
`predicted_salary`.

```python
prediction_years = 8
predicted_salary = salary_intercept + salary_slope * prediction_years
print("Predicted salary:", round(predicted_salary))
```

### Reproduce the salary prediction graph

Use the same graph structure: observations, regression line, prediction point,
equation, and dashed guides to both axes.

```python
salary_x = np.linspace(
    salary["years_experience"].min(),
    salary["years_experience"].max(),
    100,
)

plt.scatter(salary["years_experience"], salary["salary_aud"], label="Salary data")
plt.plot(
    salary_x,
    salary_intercept + salary_slope * salary_x,
    label="Regression line",
)
plt.scatter([prediction_years], [predicted_salary])
plt.plot(
    [prediction_years, prediction_years],
    [salary["salary_aud"].min(), predicted_salary],
    "--",
)
plt.plot(
    [salary["years_experience"].min(), prediction_years],
    [predicted_salary, predicted_salary],
    "--",
)
plt.text(
    0.04,
    0.95,
    "salary = 41,307 + 2,709 × years",
    transform=plt.gca().transAxes,
    va="top",
)
plt.xlabel("Years of experience")
plt.ylabel("Salary (AUD)")
plt.legend()
plt.show()
```

### Activity: Predict two salaries

Predict salary at 5 and 12 years of experience. Calculate the difference and check
it against seven years multiplied by the slope.

```python
salary_5 = salary_intercept + salary_slope * 5
salary_12 = salary_intercept + salary_slope * 12
difference = salary_12 - salary_5

print("5 years:", round(salary_5))
print("12 years:", round(salary_12))
print("Difference:", round(difference))
print("Seven slopes:", round(7 * salary_slope))
```
