## 1. Load the two datasets

Run this cell before starting the activities.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.linear_model import LinearRegression

sns.set_theme(style="whitegrid", palette="colorblind")

DATA_FOLDER = Path("../data")
HABITS_FILE = DATA_FOLDER / "student_habits.csv"
SALARY_FILE = DATA_FOLDER / "experience_and_salary.csv"

habits = pd.read_csv(HABITS_FILE)
salary = pd.read_csv(SALARY_FILE)

habits.head()
```

## 2. Fit the study-hours model

Use `np.polyfit()` to find the slope and intercept for `study_hours` and
`exam_score`. A straight line has degree 1.

```python
study_slope, study_intercept = np.polyfit(
    habits["study_hours"],
    habits["exam_score"],
    1,
)

print("Slope:", round(study_slope, 2))
print("Intercept:", round(study_intercept, 2))
```

### Check the same model with scikit-learn

Scikit-learn expects the input column as a two-dimensional table, so keep the double
brackets in `habits[["study_hours"]]`.

```python
study_X = habits[["study_hours"]]
study_y = habits["exam_score"]

study_model = LinearRegression()
study_model.fit(study_X, study_y)

print("Slope:", round(study_model.coef_[0], 2))
print("Intercept:", round(study_model.intercept_, 2))
```

**Model answer:** Yes. Both methods produce a slope of about 4.75 and an intercept of about 54.59 because they fit the same straight-line model to the same data.

## 3. Reproduce the study-hours chart

Use `sns.lmplot()` to reproduce Figure 1 from Topic 02. Hide the confidence interval
so the dots and line remain clear.

```python
sns.lmplot(
    data=habits,
    x="study_hours",
    y="exam_score",
    ci=None,
)
plt.show()
```

### Activity: Which line is better?

Overlay the original student data with two candidate lines, as shown in Figure 2.
Line A should use the fitted slope and intercept. Move Line B 12 marks above Line A.

```python
study_x = np.array([
    habits["study_hours"].min(),
    habits["study_hours"].max(),
])
best_line = study_intercept + study_slope * study_x
poor_line = best_line + 12

plt.scatter(habits["study_hours"], habits["exam_score"])
plt.plot(study_x, best_line, label="Line A")
plt.plot(study_x, poor_line, label="Line B")
plt.xlabel("Study hours")
plt.ylabel("Exam score")
plt.legend()
plt.show()
```

**Model answer:** The fitted line runs through the centre of the point cloud and keeps the vertical gaps small overall. A line above most dots would make larger errors for most students.

## 4. Fit and graph the salary model

Repeat the NumPy method with `years_experience` and `salary_aud`.

```python
salary_slope, salary_intercept = np.polyfit(
    salary["years_experience"],
    salary["salary_aud"],
    1,
)

print("Slope:", round(salary_slope, 2))
print("Intercept:", round(salary_intercept, 2))
```

### Reproduce the salary chart

Use `sns.lmplot()` to reproduce Figure 3 from Topic 02.

```python
sns.lmplot(
    data=salary,
    x="years_experience",
    y="salary_aud",
    ci=None,
)
plt.show()
```

## 5. Activity: Put the slopes into words

Explain what each slope means in context. Use **is associated with**, not
**causes**.

**Model answer:** Each additional study hour is associated with about 4.75 more exam marks. Each additional year of experience is associated with about $2,709 more salary.

## 6. Activity: Does the intercept make sense?

Consider the study model, the salary model and the height example from Topic 02.
Which intercepts are useful real-world baselines, and which mainly position a line?

**Model answer:** The study and salary intercepts are understandable baselines, but zero is just outside both observed ranges, so they should be used cautiously. The height-model intercept mainly positions the line because a height of zero centimetres is not meaningful for a student.

## 7. Assess the student-habits fit

Topic 04 introduces residuals and R². Calculate the predicted score and residual for every row, then reproduce the residual plot.

```python
habits["predicted_score"] = study_slope * habits["study_hours"] + study_intercept
habits["residual"] = habits["exam_score"] - habits["predicted_score"]

plt.scatter(habits["study_hours"], habits["residual"], alpha=0.65)
plt.axhline(0, color="black", linestyle="--")
plt.xlabel("Study hours per day")
plt.ylabel("Residual (actual score - predicted score)")
plt.title("Residuals from the student-habits linear model")
plt.show()
```

### Activity: Compare R-squared values

Use each scikit-learn model's `.score()` method. Interpret each result as the percentage of outcome variation explained by its linear relationship with the predictor.

```python
study_r_squared = study_model.score(study_X, study_y)

salary_X = salary[["years_experience"]]
salary_y = salary["salary_aud"]
salary_model = LinearRegression()
salary_model.fit(salary_X, salary_y)
salary_r_squared = salary_model.score(salary_X, salary_y)

print(f"Student habits R-squared: {study_r_squared:.3f}")
print(f"Salary R-squared: {salary_r_squared:.3f}")
```

**Answer:** The salary model has the stronger linear fit. Its points sit more tightly around the line and its R² is about 0.923, compared with about 0.849 for the student-habits model.
