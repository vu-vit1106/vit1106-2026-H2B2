````python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

sns.set_theme(style="whitegrid", palette="colorblind")

DATA_FILE = Path("data") / "diabetes.xlsx"

df = pd.read_excel(DATA_FILE, sheet_name="diabetes")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
df.head()
````

### Prepare the numeric health data

**Patient number** is numeric, but it is an identifier rather than a health measurement. **Gender** and **Diabetes** are categories, so they are outside this activity. The next cell removes the identifier and keeps the numeric columns.

````python
numeric_columns = [
    "Cholesterol",
    "Glucose",
    "HDL Chol",
    "Age",
    "Height",
    "Weight",
    "BMI",
    "Systolic BP",
    "Diastolic BP",
    "waist",
    "hip",
]

health_data = df[numeric_columns]

correlation_matrix = health_data.corr().round(2)
correlation_matrix
````

````python
plt.figure(figsize=(11, 8))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="RdBu_r",
    vmin=-1,
    vmax=1,
)
plt.title("Correlations between numeric health measurements")
plt.tight_layout()
plt.show()
````
